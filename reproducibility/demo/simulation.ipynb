# Synthetic gene-expression demo

This notebook runs bcNMF on the stored beta=8 raw-count simulation.

## 1. Setup

The root lookup works when the notebook is opened from the repository root or from `demo/`.

```python
from pathlib import Path
import sys

def repository_root():
    for candidate in (Path.cwd().resolve(), *Path.cwd().resolve().parents):
        if (candidate / "pyproject.toml").is_file() and (candidate / "src").is_dir():
            return candidate
    raise RuntimeError("Open this notebook from the bcNMF reproducibility repository.")

ROOT = repository_root()
sys.path.insert(0, str(ROOT / "src"))

import contextlib
import io
import anndata as ad
import numpy as np
import pandas as pd
import torch
sys.modules["tensorflow"] = None  # UMAP does not need TensorFlow for this analysis
import umap
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
from bcnmf import contrastive_nmf_poisson, contrastive_nmf_sse

def dense(matrix):
    return np.ascontiguousarray(matrix.toarray() if hasattr(matrix, "toarray") else matrix, dtype=np.float64)

def umap_ari(H, labels, seed=42, n_neighbors=15, min_dist=0.1):
    embedding = umap.UMAP(n_neighbors=n_neighbors, min_dist=min_dist,
                          metric="euclidean", random_state=seed).fit_transform(H.T)
    clusters = KMeans(n_clusters=len(np.unique(labels)), random_state=seed, n_init=10).fit_predict(embedding)
    return adjusted_rand_score(labels, clusters), embedding

def paper_result(dataset):
    result = pd.read_csv(ROOT / "results/paper_ari_reference.csv")
    return result[(result["dataset"] == dataset) & (result["method"] == "bcNMF")][["ari_mean", "ari_sd"]]
```

## 2. Generated data

The saved file includes target and background raw counts, target-subtype labels, background classes, and generating programs.

```python
data = np.load(ROOT / "demo/data/beta_8_seed_0.npz")
X = np.ascontiguousarray(data["X"], dtype=np.float64)
Y = np.ascontiguousarray(data["Y"], dtype=np.float64)
labels = np.asarray(data["labels"])
print(f"X={X.shape}; Y={Y.shape}; beta={float(data['beta'])}; target labels={np.bincount(labels)}")
```

## 3. bcNMF and ARI

The paper beta-series uses library-size normalized `log1p` counts and squared-error bcNMF. For beta=8, the selected setting is K=2 and alpha=2.

```python
def log_normalize(cells_by_genes):
    library_size = cells_by_genes.sum(axis=1, keepdims=True)
    library_size[library_size == 0] = 1
    return np.log1p(cells_by_genes / library_size * 1e4)

X_log = log_normalize(X.T)
Y_log = log_normalize(Y.T)
K, ALPHA, NITER, SEED = 2, 2.0, 300, 0
torch.manual_seed(20_000 + SEED * 101 + int(round(ALPHA * 1000)))
with contextlib.redirect_stdout(io.StringIO()):
    W, H_X, H_Y, _ = contrastive_nmf_sse(X_log.T, Y_log.T, K, ALPHA, niter=NITER)
clusters = KMeans(n_clusters=2, random_state=42, n_init=20).fit_predict(np.asarray(H_X).T)
ari = adjusted_rand_score(labels, clusters)
print(f"beta=8 target-subtype ARI = {ari:.4f}")
```
